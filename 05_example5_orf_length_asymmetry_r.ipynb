# Example 5 (R half): Testing Coding-Strand Asymmetry


Reads `results/example5_orf_lengths.csv` (written by this Example's own Python half — run
that notebook first) and tests, with a Wilcoxon rank-sum test, whether forward-strand ORF
lengths are significantly greater than reverse-strand ORF lengths. Produces an annotated,
log-scaled ggplot2 boxplot with the p-value printed on it.


**Why a Wilcoxon test, not a t-test:** ORF lengths are strongly right-skewed (most ORFs are
short, a few are very long), not normally distributed — exactly the situation a rank-based
test handles correctly and a t-test does not.

```r
# Test whether forward-strand ORFs are significantly longer than reverse-strand ORFs
# اختبار هل إطارات القراءة على الشريط الأمامي أطول بدلالة إحصائية من الشريط المعاكس
library(ggplot2)
dir.create("results", showWarnings = FALSE)

orf_data <- read.csv("results/example5_orf_lengths.csv")  # written by this Example's own Python half
orf_data$strand <- factor(orf_data$strand, levels = c("forward", "reverse"))

test_result <- wilcox.test(length_aa ~ strand, data = orf_data)
cat("Wilcoxon rank-sum test\n")
cat(sprintf("W = %.1f, p-value = %.2e\n", test_result$statistic, test_result$p.value))

p <- ggplot(orf_data, aes(x = strand, y = length_aa, fill = strand)) +
  geom_boxplot(alpha = 0.7, outlier.size = 0.8) +
  scale_y_log10() +
  labs(
    title = "ORF length by strand (SARS-CoV-2, NC_045512.2)",
    subtitle = sprintf("Wilcoxon rank-sum test: p = %.2e", test_result$p.value),
    x = "Strand", y = "ORF length, aa (log scale)"
  ) +
  theme_minimal() +
  theme(legend.position = "none")

ggsave("results/example5_orf_length_boxplot.png", p, width = 6, height = 4.5, dpi = 150)
cat("Saved: results/example5_orf_length_boxplot.png\n")
```
